```python
import pandas as pd
import numpy as np

data = [
    ["2021", "East", 4.5],
    ["2021", "West", 8.0],
    ["2022", "East", 12.5],
    ["2022", "West", 6.0],
    ["2023", "East", 3.0],
    ["2023", "West", 15.0]
]

df = pd.DataFrame(data, columns=["Year", "Region", "Default_Rate"])

# 1. Create a pivot table showing the average default rate per region and year (wide format)
pivot_df = pd.pivot_table(df, index='Region', columns='Year', values='Default_Rate', aggfunc='mean')

# 2. Melt the pivot table into a long format
result = pivot_df.reset_index().melt(id_vars='Region', var_name='Year', value_name='Default_Rate')

# Convert 'Year' column to integer type for proper sorting and consistency
result['Year'] = result['Year'].astype(int)

# 3. Bin the default rates into categories: Low (< 5%), Medium (5%–10%), High (> 10%)
conditions = [
    result['Default_Rate'] < 5,
    (result['Default_Rate'] >= 5) & (result['Default_Rate'] <= 10),
    result['Default_Rate'] > 10
]
choices = ['Low', 'Medium', 'High']
result['Risk_level'] = np.select(conditions, choices, default='Unknown')

# Sort the results by 'Region' and then by 'Year' as required by the sample output
result = result.sort_values(by=['Region', 'Year']).reset_index(drop=True)

# Print the final result with the desired column order
print(result[['Year', 'Region', 'Default_Rate', 'Risk_level']])
```

```python
import numpy as np

matrix_data = [
    [1, -2, 3],
    [-1, 4, 2],
    [0, -3, 5]
]

matrix = np.array(matrix_data)
print("Original Matrix:")
print(matrix)

# Replace negative numbers with 0
matrix[matrix < 0] = 0

print("Matrix after replacing negative numbers with 0:")
print(matrix)
```

```python
row_sums = matrix.sum(axis=1)
print("Row sums before normalization:")
print(row_sums)
```

```python
# To avoid division by zero for rows that sum to 0, replace 0 with 1 for calculation (this won't affect the original 0 sum rows as they will remain 0 after division if their elements are 0).
row_sums[row_sums == 0] = 1 # Temporary fix for division by zero

normalized_matrix = matrix / row_sums[:, np.newaxis]

print("Normalized Matrix (each row sums to 1):")
print(normalized_matrix)
print(normalized_matrix > 0.5)
```

### Plotting a Line Chart with Seaborn

To create a line chart, you typically need two numerical variables: one for the x-axis (often representing time or an ordered category) and one for the y-axis (representing a continuous measure).

Here's an example using some simple data:

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Sample data for a line plot
data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'electronic': [100, 120, 150, 130, 160, 180],
    'clothing': [100, 120, 150, 130, 160, 180],
    'Home decor': [100, 120, 150, 130, 160, 180]
}
sales_df = pd.DataFrame(data)

# Melt the DataFrame to long format for plotting multiple lines
sales_melted_df = sales_df.melt(id_vars=['Month'], var_name='Category', value_name='Sales')

# Create the line plot
plt.figure(figsize=(10, 6))
sns.lineplot(x='Month', y='Sales', hue='Category', data=sales_melted_df, marker='o')

plt.title('Monthly Sales Trend by Category')
plt.xlabel('Month')
plt.ylabel('Sales Revenue')
plt.grid(True)
plt.legend(title='Product Category')
plt.show()
```

```python
class Solution:
    def solve(self, A):
        if A < 0:
            # Factorial is not defined for negative numbers, or usually 0! = 1
            # Assuming for non-negative integers as per standard factorial definition.
            return 0 # Or raise an error
        elif A == 0:
            return 1
        else:
            return A * self.solve(A - 1)
```

```python
from scipy.stats import lognorm
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# --- Example 1: Varying s_shape with constant scale ---
scale_val = np.exp(1) # Keep scale constant

# Different s_shape values
s_shape_low = 0.5
s_shape_high = 1.5

data_low_s = lognorm.rvs(s=s_shape_low, loc=0, scale=scale_val, size=1000)
data_high_s = lognorm.rvs(s=s_shape_high, loc=0, scale=scale_val, size=1000)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.hist(data_low_s, bins=50, density=True, alpha=0.6, color='blue', label=f's_shape={s_shape_low}')
plt.axvline(scale_val, color='blue', linestyle='dashed', linewidth=2, label=f'Median (Scale={scale_val:.2f})') # Add axvline
plt.title('Log-normal with s_shape=0.5 (Scale=e)')
plt.xlabel('Value')
plt.ylabel('Density')
plt.legend()

plt.subplot(1, 2, 2)
plt.hist(data_high_s, bins=50, density=True, alpha=0.6, color='red', label=f's_shape={s_shape_high}')
plt.axvline(scale_val, color='red', linestyle='dashed', linewidth=2, label=f'Median (Scale={scale_val:.2f})') # Add axvline
plt.title('Log-normal with s_shape=1.5 (Scale=e)')
plt.xlabel('Value')
plt.ylabel('Density')
plt.legend()

plt.tight_layout()
plt.suptitle('Effect of Varying s_shape (Spread/Skew)', y=1.02)
plt.show()
```

As you can see, increasing `s_shape` (from 0.5 to 1.5) makes the distribution wider and more skewed to the right, even when the `scale` is kept constant. The peak also shifts to the left as the distribution becomes more spread out.

```python
# --- Example 2: Varying scale with constant s_shape ---
s_shape_val = 0.8 # Keep s_shape constant

# Different scale values
scale_low = np.exp(0.5) # exp(0.5) ~ 1.65
scale_high = np.exp(1.5) # exp(1.5) ~ 4.48

data_low_scale = lognorm.rvs(s=s_shape_val, loc=0, scale=scale_low, size=1000)
data_high_scale = lognorm.rvs(s=s_shape_val, loc=0, scale=scale_high, size=1000)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.hist(data_low_scale, bins=50, density=True, alpha=0.6, color='green', label=f'scale={scale_low:.2f}')
sns.kdeplot(data_low_scale, color='darkgreen', ax=plt.gca()) # Overlay KDE
plt.axvline(scale_low, color='green', linestyle='dashed', linewidth=2, label=f'Median (Scale={scale_low:.2f})') # Add axvline
plt.title('Log-normal with Scale=1.65 (s_shape=0.8)')
plt.xlabel('Value')
plt.ylabel('Density')
plt.legend()

plt.subplot(1, 2, 2)
plt.hist(data_high_scale, bins=50, density=True, alpha=0.6, color='purple', label=f'scale={scale_high:.2f}')
sns.kdeplot(data_high_scale, color='darkred', ax=plt.gca()) # Overlay KDE
plt.axvline(scale_high, color='purple', linestyle='dashed', linewidth=2, label=f'Median (Scale={scale_high:.2f})') # Add axvline
plt.title('Log-normal with Scale=4.48 (s_shape=0.8)')
plt.xlabel('Value')
plt.ylabel('Density')
plt.legend()

plt.tight_layout()
plt.suptitle('Effect of Varying Scale (Median/Location)', y=1.02)
plt.show()
```

Here, with a constant `s_shape`, changing the `scale` (from ~1.65 to ~4.48) effectively shifts the entire distribution along the x-axis, increasing the typical values generated without significantly changing the fundamental shape or skewness. The peaks of the distributions are at higher values for a larger `scale`.
