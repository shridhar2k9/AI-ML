```python
import pandas as pd
import numpy as np

data = [
    ["2021", "East", 4.5],
    ["2021", "West", 8.0],
    ["2022", "East", 12.5],
    ["2022", "West", 6.0],
    ["2023", "East", 3.0],
    ["2023", "West", 15.0]
]

df = pd.DataFrame(data, columns=["Year", "Region", "Default_Rate"])

# 1. Create a pivot table showing the average default rate per region and year (wide format)
pivot_df = pd.pivot_table(df, index='Region', columns='Year', values='Default_Rate', aggfunc='mean')

# 2. Melt the pivot table into a long format
result = pivot_df.reset_index().melt(id_vars='Region', var_name='Year', value_name='Default_Rate')

# Convert 'Year' column to integer type for proper sorting and consistency
result['Year'] = result['Year'].astype(int)

# 3. Bin the default rates into categories: Low (< 5%), Medium (5%–10%), High (> 10%)
conditions = [
    result['Default_Rate'] < 5,
    (result['Default_Rate'] >= 5) & (result['Default_Rate'] <= 10),
    result['Default_Rate'] > 10
]
choices = ['Low', 'Medium', 'High']
result['Risk_level'] = np.select(conditions, choices, default='Unknown')

# Sort the results by 'Region' and then by 'Year' as required by the sample output
result = result.sort_values(by=['Region', 'Year']).reset_index(drop=True)

# Print the final result with the desired column order
print(result[['Year', 'Region', 'Default_Rate', 'Risk_level']])
```

```python
from scipy.stats import lognorm
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# --- Example 1: Varying s_shape with constant scale ---
scale_val = np.exp(1) # Keep scale constant

# Different s_shape values
s_shape_low = 0.5
s_shape_high = 1.5

data_low_s = lognorm.rvs(s=s_shape_low, loc=0, scale=scale_val, size=1000)
data_high_s = lognorm.rvs(s=s_shape_high, loc=0, scale=scale_val, size=1000)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.hist(data_low_s, bins=50, density=True, alpha=0.6, color='blue', label=f's_shape={s_shape_low}')
plt.axvline(scale_val, color='blue', linestyle='dashed', linewidth=2, label=f'Median (Scale={scale_val:.2f})') # Add axvline
plt.title('Log-normal with s_shape=0.5 (Scale=e)')
plt.xlabel('Value')
plt.ylabel('Density')
plt.legend()

plt.subplot(1, 2, 2)
plt.hist(data_high_s, bins=50, density=True, alpha=0.6, color='red', label=f's_shape={s_shape_high}')
plt.axvline(scale_val, color='red', linestyle='dashed', linewidth=2, label=f'Median (Scale={scale_val:.2f})') # Add axvline
plt.title('Log-normal with s_shape=1.5 (Scale=e)')
plt.xlabel('Value')
plt.ylabel('Density')
plt.legend()

plt.tight_layout()
plt.suptitle('Effect of Varying s_shape (Spread/Skew)', y=1.02)
plt.show()
```

As you can see, increasing `s_shape` (from 0.5 to 1.5) makes the distribution wider and more skewed to the right, even when the `scale` is kept constant. The peak also shifts to the left as the distribution becomes more spread out.

# Task
Simulate drawing two balls without replacement from a box containing 3 red and 2 blue balls (total 5 balls). Estimate the probabilities for two events over 10,000 simulations: 1. The second ball drawn is red. 2. Both balls drawn are of the same color. Round the estimated probabilities to two decimal places and print them.

## Set Up Simulation Parameters

### Subtask:
Define the contents of the 'box' programmatically, representing the balls, and specify the total number of simulations (N = 10,000) that will be performed.


**Reasoning**:
The subtask requires defining the contents of the 'box' and the number of simulations. I will use a Python code block to create a list representing the balls and a variable for the number of simulations.



```python
box = ['red', 'red', 'red', 'blue', 'blue']
num_simulations = 10000

print(f"Initial box contents: {box}")
print(f"Number of simulations: {num_simulations}")
```

## Simulate Drawing Two Balls

### Subtask:
Implement the simulation loop to draw two balls without replacement for each of the 10,000 simulations.


**Reasoning**:
I need to initialize the counters and then implement the simulation loop, including drawing balls without replacement and updating event counts. This directly addresses instructions 1-6.



```python
import random

second_ball_red_count = 0
both_same_color_count = 0

for _ in range(num_simulations):
    current_box = list(box) # Create a copy to simulate drawing without replacement

    # Draw first ball
    first_ball = random.choice(current_box)
    current_box.remove(first_ball)

    # Draw second ball
    second_ball = random.choice(current_box)

    # Check for event 1: Second ball is red
    if second_ball == 'red':
        second_ball_red_count += 1

    # Check for event 2: Both balls are of the same color
    if first_ball == second_ball:
        both_same_color_count += 1

print(f"Finished simulations. Second ball red count: {second_ball_red_count}")
print(f"Finished simulations. Both same color count: {both_same_color_count}")
```

**Reasoning**:
Calculate the estimated probabilities for both events, round them to two decimal places, and print them in the specified format.

```python
"""
6. Analyze the relationship between coupon usage and revenue generation. How can discount strategies be optimized to maximize revenue while maintaining profitability?
Hint: Analyze the overall revenue share from transactions involving coupons versus those without.
Hint: Consider if coupon usage impacts the value of individual transactions, not just the volume. This might involve comparing average transaction values between coupon users and non-users using a statistical test.
"""
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns

# Ensure Sales_Amount is calculated
online_sales_data['Sales_Amount'] = online_sales_data['Quantity'] * online_sales_data['Avg_Price']

# Grouping transactions
# Let's define coupon transactions: 'Used' or 'Clicked' vs 'Not Used'
coupon_active = online_sales_data[online_sales_data['Coupon_Status'].isin(['Used', 'Clicked'])]
coupon_not_used = online_sales_data[online_sales_data['Coupon_Status'] == 'Not Used']

# 1. Revenue Share Analysis
total_revenue = online_sales_data['Sales_Amount'].sum()
rev_active = coupon_active['Sales_Amount'].sum()
rev_not_used = coupon_not_used['Sales_Amount'].sum()

print("Overall Revenue Share Analysis")
print(f"Total Revenue: ${total_revenue:,.2f}")
print(f"Revenue from Transactions with Coupon Active (Clicked/Used): ${rev_active:,.2f} ({rev_active/total_revenue*100:.2f}%)")
print(f"Revenue from Transactions without Coupon (Not Used): ${rev_not_used:,.2f} ({rev_not_used/total_revenue*100:.2f}%)")
print()

# 2. Compare Individual Transaction Values
mean_active = coupon_active['Sales_Amount'].mean()
mean_not_used = coupon_not_used['Sales_Amount'].mean()

print("Average Transaction Values")
print(f"Average transaction value with Coupon (Clicked/Used): ${mean_active:.2f}")
print(f"Average transaction value without Coupon (Not Used): ${mean_not_used:.2f}")
print()

# 3. Statistical T-Test
t_stat, p_val = stats.ttest_ind(coupon_active['Sales_Amount'], coupon_not_used['Sales_Amount'], equal_var=False)
print("Statistical Significance Test (T-Test)")
print(f"T-statistic: {t_stat:.4f}")
print(f"P-value: {p_val:.4e}")
if p_val < 0.05:
    print("Result: Statistically Significant. Coupon status significantly impacts individual transaction sizes.")
else:
    print("Result: Not Statistically Significant.")
```
